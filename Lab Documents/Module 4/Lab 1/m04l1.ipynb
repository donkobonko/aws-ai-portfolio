### Title: Ecommerce Orders.

This notebook is for cleaning up the list of orders. Only alter df_clean, and run every cell in order. The rows have already been dropped from 500 to 466 due to missing unit_prices and impossible shipping_costs.

```python
import pandas as pd
df = pd.read_csv('m4_ecommerce_orders.csv')

display(df.shape)
display(df.dtypes)
display(df.isna().sum())
display(df.describe())
display(df.head())
```

```python
# the missing ones, in one line
print(df.isna().sum())

# the impossible ones, one check each. Two are done for you.
print((df['shipping_cost'] < 0).sum(), 'negative shipping costs')
print((df['quantity'] < 1).sum(), 'quantities below one')

# write one more check of your own for another column
print((df['unit_price'] < 0).sum(), 'negative unit prices')
```

```python
# run this before you decide which fix to use
before = df['quantity'].mean()
after  = df.dropna(subset=['unit_price'])['quantity'].mean() # means to drop N/A rows in that column unit_price
print(before, after)

# then apply your two fixes and save the result as df_clean

# WHAT: 21 blank unit prices
# DID: Dropped those 21 rows
# COSTS: Lost actual customer details for that order, like payment, date, category, etc.
df_clean = df.dropna(subset=['unit_price'])

# WHAT: 13  Negative shipping costs
# DID: Got rid of the rows with negative shipping costs
# COSTS: This also lost 13 real order details
df_clean = df_clean[df_clean['shipping_cost'] >= 0] # gets rid of rows with negative values

print(df_clean.shape)
display(df_clean.head())
print(before, df_clean['quantity'].mean())
```

```python
# the pattern: group by one column, average another
print(df_clean.groupby('category')['quantity'].mean().sort_values(ascending=False))

print()

# now do the same for region and shipping_cost
print(df_clean.groupby('region')['shipping_cost'].mean().sort_values(ascending=False))
```
